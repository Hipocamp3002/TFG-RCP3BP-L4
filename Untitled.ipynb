```julia
using Plots
```

```julia
function hamiltonian(u,mu)
    q1,q2,p1,p2 = u

    return (p1^2 + p2^2)/2.0 - q1*p2 + q2*p1 - (1-mu)/sqrt((q1+mu)^2 + q2^2) - mu/sqrt((q1-1+mu)^2 + q2^2)
end

mu = 0.1
radius = 0.3
height = 5
L4 = [0.5-mu,sqrt(3)/2,-sqrt(3)/2,0.5-mu]
H_L4 = hamiltonian(L4,mu)
```

```julia
dL1 = (mu/3)^(1/3)
L1_eq(u) = u^5 - (3-mu)*u^4 + (3-2mu)*u^3 - mu*u^2 + 2mu*u - mu
L1_de(u) = 5u^4 - 4*(3-mu)*u^3 + 3*(3-2mu)*u^2 - 2*mu*u + 2mu
for i = 1:16
    dL1 = dL1 - L1_eq(dL1)/L1_de(dL1)
end

L1 = [(1-mu)-dL1,0,0,(1-mu)-dL1]
H_L1 = hamiltonian(L1,mu)

dL2 = (mu/3)^(1/3)
L2_eq(u) = u^5 + (3-mu)*u^4 + (3-2mu)*u^3 - mu*u^2 - 2mu*u - mu
L2_de(u) = 5u^4 + 4*(3-mu)*u^3 + 3*(3-2mu)*u^2 - 2*mu*u - 2mu
for i = 1:16
    dL2 = dL2 - L2_eq(dL2)/L2_de(dL2)
end

L2 = [(1-mu)+dL2,0,0,(1-mu)+dL2]
H_L2 = hamiltonian(L2,mu)

dL3 = (mu/3)^(1/3)
L3_eq(u) = u^5 + (3-mu)*u^4 + (3-2mu)*u^3 - mu*u^2 - 2mu*u - mu
L3_de(u) = 5u^4 + 4*(3-mu)*u^3 + 3*(3-2mu)*u^2 - 2*mu*u - 2mu
for i = 1:16
    dL3 = dL3 - L3_eq(dL3)/L2_de(dL3)
end

L3 = [-mu-dL3,0,0,-mu-dL3]
H_L3 = hamiltonian(L3,mu)
@show L3
```

```julia
resolution = 500
ang_range = LinRange(0,2pi,resolution) |> collect
lin_range = LinRange(-height,height,resolution) |> collect
```

```julia
Q1 = lin_range*0.5
Q2 = lin_range*0.5

ang = 1
@gif for d in -2:0.1:2
points = Array{Float64,2}(undef,resolution,resolution)
for (i,q1) in enumerate(Q1)
    for (j,q2) in enumerate(Q2)
        points[j,i] = log(abs(hamiltonian([q1,q2,-q2+sin(ang)*d,q1+cos(ang)*d],mu) - H_L4))
    end
end

heatmap(Q1,Q2,points)
end
```

```julia
#Q section
Q = [cos.(ang_range) sin.(ang_range)] * radius
P = lin_range

Q = mapslices((x) -> x + L4[1:2], Q,dims = 2)
#P = mapslices((x) -> x + L4[3:4], P,dims = 1)

points = Array{Vector{Float64},2}(undef,resolution,resolution)
for (i,q) in enumerate(eachrow(Q))
    for (j,p) in enumerate(P)
        points[i,j] = [q;p;-1.1] - L4
    end
end

heatmap(lin_range,ang_range,log.(abs.(hamiltonian.(points,mu) .- H_L4)))
```

```julia
#P section
P = [cos.(ang_range) sin.(ang_range)] * radius
Q = lin_range


points = Array{Vector{Float64},2}(undef,resolution,resolution)
for (i,q) in enumerate(Q)
    for (j,p) in enumerate(eachrow(P))
        points[i,j] = [q;0.0;p] - L4
    end
end

heatmap(ang_range,lin_range,log.(abs.(hamiltonian.(points,mu) .- H_L4)))
```

```julia
#QP section
Q = [cos.(ang_range) sin.(ang_range)] * radius
P = [cos.(ang_range) sin.(ang_range)] * radius

points = Array{Vector{Float64},2}(undef,resolution,resolution)
for (i,q) in enumerate(eachrow(Q))
    for (j,p) in enumerate(eachrow(P))
        points[i,j] = [q;p] - L4
    end
end

heatmap(ang_range,ang_range,log.(abs.(hamiltonian.(points,mu) .- H_L4)))
```

```julia
#true spherical
theta = deg2rad(0)
angles = [vcat.(a,ang_range) for a in ang_range]

points = Array{Vector{Float64},2}(undef,resolution,resolution)
for (i,row) in enumerate(angles)
    for (j, angs) in enumerate(row)
        a1 ,a2 = angs
        p = [cos(a1), sin(a1)*cos(a2), sin(a1)*sin(a2)*cos(theta), sin(a1)*sin(a2)*sin(theta)]
        points[i,j] = p * radius - L4
    end
end

heatmap(ang_range,ang_range,log.(abs.(hamiltonian.(points,mu) .- H_L4)))
```

```julia
theta = deg2rad(0)
pos = [vcat.(a,lin_range) for a in lin_range]

points = Array{Vector{Float64},2}(undef,resolution,resolution)
@gif for theta in collect(LinRange(0,2pi,50))
for (i,row) in enumerate(pos)
    for (j, cur_pos) in enumerate(row)
        p1 ,p2 = cur_pos
        point = [cos(theta)*radius,sin(theta)*radius,p1,p2]
        points[i,j] = point - L4
    end
end
@show theta
heatmap(lin_range,lin_range,log.(abs.(hamiltonian.(points,mu) .- H_L4)))
end
```
