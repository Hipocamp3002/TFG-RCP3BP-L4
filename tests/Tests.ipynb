```julia

using Plots
using LinearAlgebra
using DiffEqPhysics
include("RCP3BP.jl")
```

Probando valores propios

```julia
mu = 0.4
vaps = eigenValues(mu)
veps = eigenVectors(mu)
A = X_H(mu)
for i = 1:4
    println(repr("text/plain",A*veps[i] - vaps[i]*veps[i]))
end
veps
```

Trobar coeficients $C_0$ i $C_1$ que eliminin part imaginària.

$C_0 - C_1 = 0$ 

Suposem $C_0 = cos(\theta), C_1 = sin(\theta)$, aleshores,$\theta = \pi / 4$

```julia
theta = π/4

a = real(vaps[1])
b = imag(vaps[1])

v1 = veps[1]
v2 = veps[2]

tf = 2π/b

e_at = exp(a*tf)

expected_abs_error = 0.1^10
eps = 1.0


curr_abs_error = 10
while(curr_abs_error>expected_abs_error)
    eps = eps/2.0
    C0 = cos(theta)*eps
    C1 = sin(theta)*eps

    pos_ini = L4(mu)+real(C0*v1 + C1*v2)
    aprox_final = L4(mu)+real(e_at*C0*v1 + e_at*C1*v2)

    param = (mu)
    tspan = (0.0,tf)
    prob = ODEProblem(orbit!,pos_ini,tspan,param)
    sol = solve(prob,Feagin14(),dtmax=0.1)
    pos_final = sol.u[end]

    curr_abs_error=norm(aprox_final-pos_final)
end
println(eps)
```

Graficar resultad

```julia
C0 = cos(theta)*eps
C1 = sin(theta)*eps

pos_ini = L4(mu)+real(C0*v1 + C1*v2)
aprox_final = L4(mu)+real(e_at*C0*v1 + e_at*C1*v2)

param = (mu)
tspan = (0.0,tf)
prob = ODEProblem(orbit!,pos_ini,tspan,param)
sol = solve(prob,Feagin14(),dtmax=0.1)

plot(sol,idxs=(1,2),label="Integracio")
plot!([pos_ini[1],aprox_final[1]],[pos_ini[2],aprox_final[2]],label="Linealitzacio")
savefig("epsilon.svg")
```
