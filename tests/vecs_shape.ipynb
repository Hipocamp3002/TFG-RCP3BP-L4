```julia
using Plots
include("RCP3BP.jl")
```

```julia
function f(v)
    v1,v2,_,_ = v[1]
    u1,u2,_,_ = v[2]

    b1 = 1/(u1-u2*(v1/v2))
    a1 = (1-u1*b1)/v1

    b2 = 1/(u2-u1*(v2/v1))
    a2 = (1-u2*b2)/v2

    return [a1*v[1]+b1*v[2], a2*v[1]+b2*v[2]]
end


Ivecs = []
Evecs = []
for mu in 0.05:0.01:0.5
    s = Sistem(mu)
    push!(Ivecs,f(s.Ivecs))
    push!(Evecs,s.Evecs)
end
```

```julia
x = []
y = []
for ang in 1:360
    theta = deg2rad(ang)
    u = Ivecs[1][1]*cos(theta) + Ivecs[1][2]*sin(theta)
    push!(x,u[1])
    push!(y,u[2])
end
plot(x,y;aspect_ratio = :equal)
```
