```julia
using CSV
using GZip
using ProgressBars
```

```julia
file_paths = []
for (path,dirs,files) in walkdir("data")
    for file in files
       if file[end-3:end] == ".csv"
           push!(file_paths, (path,file))
       end
    end
end
file_paths[2]
```

```julia
n=67
path, file = rand(file_paths)
s1 = filesize(path*"/"*file)
```

```julia
data = CSV.File(path*"/"*file)
touch("out")
@time GZip.open("out","w") do f
    for (uStr,t,theta) in data
        u = parse.(Float64,split(uStr[2:end-1],","))
        write(f,u)
    end
end
s2 = filesize("out")
```

```julia
s1/s2
```

```julia
GZip.open("out","r") do f
    while !eof(f)
        y = Vector{Float64}(undef,4)
        read!(f,y)
        @show y
    end
end
```
